```python
# Gen AI - Assignment 3: Comprehensive Visualizations (The Ultimate Collection)
# Dataset: Breast Cancer Wisconsin (Diagnostic)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer

# Load and prepare the dataset
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['diagnosis'] = pd.Series(data.target).map({0: 'Malignant', 1: 'Benign'})
display(df.head())
```

```python
# --- MATPLOTLIB VISUALIZATIONS (Plots 1-20) ---
plt.style.use('default')
mal = df[df['diagnosis'] == 'Malignant']
ben = df[df['diagnosis'] == 'Benign']
```

```python
# 1. Line Plot: Tracking trend of sorted 'mean area'
plt.figure(figsize=(8, 4))
plt.plot(np.sort(df['mean area']), color='blue', linewidth=2)
plt.title('Line Plot: Sorted Mean Area Trend')
plt.show()
```

```python
# 2. Basic Scatter Plot: Analyzing correlation between radius and texture
plt.figure(figsize=(8, 4))
plt.scatter(mal['mean radius'], mal['mean texture'], c='red', label='Malignant', alpha=0.6, edgecolors='k')
plt.scatter(ben['mean radius'], ben['mean texture'], c='green', label='Benign', alpha=0.6, edgecolors='k')
plt.title('Scatter Plot: Basic')
plt.legend()
plt.show()
```

```python
# 3. Advanced Scatter Plot: Size reflects Area, Color reflects Smoothness
plt.figure(figsize=(8, 4))
scatter = plt.scatter(df['mean radius'], df['mean texture'], 
                      c=df['mean smoothness'], s=df['mean area']*0.15, 
                      alpha=0.7, cmap='viridis', edgecolors='white', linewidth=0.5)
plt.colorbar(scatter, label='Smoothness')
plt.title('Scatter: Size=Area, Color=Smoothness')
plt.show()
```

```python
# 4. 3D Scatter Plot: Multi-dimensional feature space
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(mal['mean radius'], mal['mean texture'], mal['mean concavity'], c='r', alpha=0.6, label='Malignant', edgecolors='w')
ax.scatter(ben['mean radius'], ben['mean texture'], ben['mean concavity'], c='g', alpha=0.6, label='Benign', edgecolors='w')
ax.set(xlabel='Radius', ylabel='Texture', zlabel='Concavity', title='3D Scatter Plot')
ax.legend()
plt.show()
```

```python
# 5. Bar Plot: Comparing average mean radius with proper spacing
plt.figure(figsize=(8, 4))
means = [mal['mean radius'].mean(), ben['mean radius'].mean()]
labels = ['Malignant', 'Benign']
x_pos = np.arange(len(labels))
plt.bar(x_pos, means, color=['#e74c3c', '#2ecc71'], width=0.5, edgecolor='black', linewidth=1.2)
plt.xticks(x_pos, labels)
plt.title('Bar Plot: Average Mean Radius')
plt.show()
```

```python
# 6. Horizontal Bar Plot: Top 5 samples by 'mean area' with spacing
plt.figure(figsize=(8, 4))
top5 = df.nlargest(5, 'mean area')
y_pos = np.arange(len(top5))
plt.barh(y_pos, top5['mean area'], color='orange', height=0.6, edgecolor='black')
plt.yticks(y_pos, [f"Sample {i}" for i in top5.index])
plt.title('Horizontal Bar: Top 5 Mean Areas')
plt.show()
```

```python
# 7. Stacked Bar Plot: Radius and Texture combined
plt.figure(figsize=(8, 4))
mean_rad = [mal['mean radius'].mean(), ben['mean radius'].mean()]
mean_tex = [mal['mean texture'].mean(), ben['mean texture'].mean()]
plt.bar(np.arange(2), mean_rad, color='#e74c3c', width=0.5, edgecolor='black', label='Radius')
plt.bar(np.arange(2), mean_tex, bottom=mean_rad, color='#3498db', width=0.5, edgecolor='black', label='Texture')
plt.xticks(np.arange(2), ['Malignant', 'Benign'])
plt.title('Stacked Bar Plot: Radius & Texture')
plt.legend()
plt.show()
```

```python
# 8. Basic Histogram: Distribution of mean smoothness
plt.figure(figsize=(8, 4))
plt.hist(df['mean smoothness'], bins=20, color='purple', alpha=0.7, rwidth=0.85, edgecolor='black')
plt.title('Histogram: Basic')
plt.show()
```

```python
# 9. Overlapping Histogram: Comparing Malignant vs Benign
plt.figure(figsize=(8, 4))
plt.hist(mal['mean smoothness'], bins=20, alpha=0.6, label='Malignant', color='red', rwidth=0.9)
plt.hist(ben['mean smoothness'], bins=20, alpha=0.6, label='Benign', color='green', rwidth=0.9)
plt.title('Histogram: Overlapping')
plt.legend()
plt.show()
```

```python
# 10. Stacked Histogram
plt.figure(figsize=(8, 4))
plt.hist([mal['mean smoothness'], ben['mean smoothness']], bins=20, stacked=True, label=['Malignant', 'Benign'], edgecolor='k')
plt.title('Histogram: Stacked')
plt.legend()
plt.show()
```

```python
# 11. Box Plot: Analyzing spread and outliers
plt.figure(figsize=(8, 4))
plt.boxplot([mal['mean perimeter'], ben['mean perimeter']], labels=['Malignant', 'Benign'], patch_artist=True)
plt.title('Box Plot: Mean Perimeter')
plt.show()
```

```python
# 12. Violin Plot (MPL): Density distribution
plt.figure(figsize=(8, 4))
plt.violinplot([mal['mean area'], ben['mean area']])
plt.xticks([1, 2], ['Malignant', 'Benign'])
plt.title('Violin Plot (MPL)')
plt.show()
```

```python
# 13. Pie Chart: Diagnosis breakdown
plt.figure(figsize=(6, 6))
counts = df['diagnosis'].value_counts()
plt.pie(counts, labels=counts.index, autopct='%1.1f%%', colors=['#2ecc71', '#e74c3c'], 
        explode=(0.05, 0), shadow=True, startangle=90, wedgeprops={'edgecolor': 'black'})
plt.title('Pie Chart: Diagnosis Breakdown')
plt.show()
```

```python
# 14. Step Plot: Trend mapping for mean radius
plt.figure(figsize=(8, 4))
plt.step(range(50), df['mean radius'].head(50), color='cyan', where='mid')
plt.title('Step Plot: Mean Radius')
plt.show()
```

```python
# 15. Hexbin Plot: 2D distribution density
plt.figure(figsize=(8, 4))
plt.hexbin(df['mean radius'], df['mean area'], gridsize=20, cmap='Blues')
plt.colorbar(label='Count')
plt.title('Hexbin Plot: Radius vs Area')
plt.show()
```

```python
# 16. Error Bar Plot: Visualizing mean and std deviation
plt.figure(figsize=(8, 4))
means = [mal['mean texture'].mean(), ben['mean texture'].mean()]
stds = [mal['mean texture'].std(), ben['mean texture'].std()]
plt.errorbar(['Malignant', 'Benign'], means, yerr=stds, fmt='o', color='purple', ecolor='gray', capsize=5, markersize=10)
plt.title('Error Bar Plot: Mean Texture with Std Dev')
plt.show()
```

```python
# 17. Fill Between: Highlighting variance/spread
plt.figure(figsize=(8, 4))
subset = df.head(50)
x = np.arange(50)
plt.plot(x, subset['mean area'], color='teal')
plt.fill_between(x, subset['mean area'] - 50, subset['mean area'] + 50, color='teal', alpha=0.3)
plt.title('Fill Between Plot: Mean Area (First 50 samples)')
plt.show()
```

```python
# 18. Stem Plot: Emphasizing discrete values
plt.figure(figsize=(8, 4))
plt.stem(df['mean concavity'].head(40), linefmt='grey', markerfmt='D', basefmt='r-')
plt.title('Stem Plot: Mean Concavity (First 40)')
plt.show()
```

```python
# 19. 2D Histogram: Density representation
plt.figure(figsize=(8, 4))
plt.hist2d(df['mean radius'], df['mean perimeter'], bins=30, cmap='plasma')
plt.colorbar(label='Density')
plt.title('2D Histogram: Radius vs Perimeter')
plt.show()
```

```python
# 20. Contour Plot (Tricontourf): Mapping a 3rd dimension on 2D space
plt.figure(figsize=(8, 4))
contour = plt.tricontourf(df['mean radius'], df['mean texture'], df['mean area'], levels=15, cmap='coolwarm')
plt.colorbar(contour, label='Mean Area')
plt.title('Contour Plot: Radius, Texture, Area Mapping')
plt.show()
```

```python
# --- SEABORN VISUALIZATIONS (Plots 21-40) ---
sns.set_theme(style="whitegrid", palette="deep")
diag_palette = {'Malignant':'red', 'Benign':'green'}
```

```python
# 21. SNS Vertical Bar Plot: Mean area with error bars
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x='diagnosis', y='mean area', palette=diag_palette, edgecolor='black')
plt.title('SNS Vertical Bar Plot')
plt.show()
```

```python
# 22. SNS Horizontal Bar Plot
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x='mean area', y='diagnosis', palette=diag_palette, edgecolor='black')
plt.title('SNS Horizontal Bar Plot')
plt.show()
```

```python
# 23. SNS Grouped (Dodged) Bar Plot
plt.figure(figsize=(8, 5))
melted_bars = pd.melt(df, id_vars=['diagnosis'], value_vars=['mean radius', 'mean texture'])
sns.barplot(data=melted_bars, x='variable', y='value', hue='diagnosis', palette=diag_palette, edgecolor='black')
plt.title('SNS Grouped Bar Plot')
plt.show()
```

```python
# 24. SNS Count Plot: Frequency count
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='diagnosis', palette=diag_palette, edgecolor='black')
plt.title('SNS Count Plot')
plt.show()
```

```python
# 25. SNS Scatter Plot: Basic Hue mapping
plt.figure(figsize=(8, 4))
sns.scatterplot(data=df, x='mean radius', y='mean compactness', hue='diagnosis', palette=diag_palette)
plt.title('SNS Scatter: Basic Hue')
plt.show()
```

```python
# 26. SNS Advanced Scatter: Hue, Size, and Style
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='mean radius', y='mean compactness', hue='diagnosis', 
                size='mean area', sizes=(40, 300), style='diagnosis', alpha=0.75, 
                palette=diag_palette, edgecolor='black')
plt.title('SNS Scatter: Advanced Variables')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# 27. SNS Regression Plot (lmplot)
sns.lmplot(data=df, x='mean radius', y='mean area', hue='diagnosis', 
           palette=diag_palette, height=5, aspect=1.5, markers=['o', 's'])
plt.title('SNS lmplot')
plt.show()
```

```python
# 28. SNS KDE Plot: 1D Density
plt.figure(figsize=(8, 4))
sns.kdeplot(data=df, x='mean radius', hue='diagnosis', fill=True, palette=diag_palette, alpha=0.5)
plt.title('SNS KDE Plot')
plt.show()
```

```python
# 29. SNS Advanced KDE: 2D Density overlaid with scatter
plt.figure(figsize=(8, 6))
sns.kdeplot(data=df, x="mean radius", y="mean texture", hue="diagnosis", fill=True, alpha=0.6, levels=6, palette=diag_palette)
sns.scatterplot(data=df, x="mean radius", y="mean texture", hue="diagnosis", s=15, alpha=0.8, legend=False, palette={'Malignant':'darkred', 'Benign':'darkgreen'})
plt.title('SNS 2D KDE with Scatter Overlay')
plt.show()
```

```python
# 30. SNS Basic Box Plot
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x='diagnosis', y='mean area', palette=diag_palette)
plt.title('SNS Box Plot')
plt.show()
```

```python
# 31. SNS Boxen Plot: Enhanced Outliers
plt.figure(figsize=(8, 4))
sns.boxenplot(data=df, x='diagnosis', y='mean area', palette=diag_palette)
plt.title('SNS Boxen Plot')
plt.show()
```

```python
# 32. SNS Violin Plot (Basic)
plt.figure(figsize=(8, 4))
sns.violinplot(data=df, x='diagnosis', y='mean symmetry', palette=diag_palette)
plt.title('SNS Violin Plot')
plt.show()
```

```python
# 33. SNS Swarm & Violin Layered Plot
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='diagnosis', y='mean symmetry', palette=diag_palette, inner=None, linewidth=0)
sns.swarmplot(data=df, x='diagnosis', y='mean symmetry', color='black', alpha=0.6, size=4)
plt.title('SNS Swarm Layered over Violin')
plt.show()
```

```python
# 34. SNS Joint Plot (Hexbin)
sns.jointplot(data=df, x='mean compactness', y='mean concavity', kind='hex', color='#2ecc71', height=6)
plt.suptitle('SNS Joint Plot (Hexbin)', y=1.02)
plt.show()
```

```python
# 35. SNS Pair Plot
subset_cols = ['mean radius', 'mean texture', 'mean perimeter', 'diagnosis']
sns.pairplot(df[subset_cols], hue='diagnosis', palette=diag_palette, markers=["o", "s"], height=2.5)
plt.suptitle('SNS Pair Plot', y=1.02)
plt.show()
```

```python
# 36. SNS Strip + Point Plot Layered
plt.figure(figsize=(8, 5))
sns.stripplot(data=df, x='diagnosis', y='mean perimeter', alpha=0.3, jitter=True, palette=diag_palette)
sns.pointplot(data=df, x='diagnosis', y='mean perimeter', color='black', markers="D", capsize=.1)
plt.title('SNS Strip + Point Plot')
plt.show()
```

```python
# 37. SNS Masked Correlation Heatmap
plt.figure(figsize=(8, 6))
mean_cols = [c for c in df.columns if 'mean' in c]
corr = df[mean_cols].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5)
plt.title('SNS Masked Heatmap')
plt.show()
```

```python
# 38. SNS Ridge Plot (FacetGrid)
melted_df = pd.melt(df, id_vars=['diagnosis'], value_vars=['mean radius', 'mean texture', 'mean perimeter'])
g = sns.FacetGrid(melted_df, row="variable", hue="diagnosis", aspect=4, height=1.5, palette=diag_palette)
g.map(sns.kdeplot, "value", fill=True, alpha=0.6, linewidth=1.5)
g.map(sns.kdeplot, "value", color="w", linewidth=2)
g.add_legend()
g.fig.suptitle('SNS Ridge Plot', y=1.02)
plt.show()
```

```python
# 39. SNS ECDF Plot
plt.figure(figsize=(8, 4))
sns.ecdfplot(data=df, x='mean area', hue='diagnosis', palette=diag_palette, linewidth=2)
plt.title('SNS ECDF Plot')
plt.show()
```

```python
# 40. SNS Clustermap
# Taking a subset of data (mean features of first 50 samples) to visualize clustering cleanly
subset_df = df[[c for c in df.columns if 'mean' in c]].head(50)
sns.clustermap(subset_df, standard_scale=1, cmap='mako', figsize=(10, 8))
plt.suptitle('SNS Clustermap', y=1.02)
plt.show()
```
