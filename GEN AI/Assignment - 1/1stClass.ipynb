```python
import tensorflow as tf
```

```python
scalar = tf.constant(10)
print("Scalar:")
print(scalar)
```

```python
vector = tf.constant([1, 2, 3, 4, 5])
print("Vector:")
print(vector)
```

```python
matrix = tf.constant([[1, 2],
                      [3, 4]])
print("Matrix:")
print(matrix)
```

```python
tensor3d = tf.constant([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print("3-D Tensor:")
print(tensor3d)
```

```python
# Create a tensor
tensor = tf.constant([[10, 20, 30],
                      [40, 50, 60]])

# Display tensor properties
print("Tensor:")
print(tensor)

print("\nShape:", tensor.shape)
print("Data Type:", tensor.dtype)
print("Dimensions:", tensor.ndim)
print("Number of Elements:", tf.size(tensor).numpy())
```

```python
# Create two tensors
A = tf.constant([10, 20, 30])
B = tf.constant([1, 2, 3])

# Arithmetic Operations
print("Tensor A:", A)
print("Tensor B:", B)

print("\nAddition:")
print(tf.add(A, B))

print("\nSubtraction:")
print(tf.subtract(A, B))

print("\nMultiplication:")
print(tf.multiply(A, B))

print("\nDivision:")
print(tf.divide(A, B))
```

```python
# Create two matrices
A = tf.constant([[1, 2],
                 [3, 4]])

B = tf.constant([[5, 6],
                 [7, 8]])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

# Matrix Addition
print("\nMatrix Addition:")
print(tf.add(A, B))

# Matrix Multiplication
print("\nMatrix Multiplication:")
print(tf.matmul(A, B))

# Transpose of Matrix A
print("\nTranspose of Matrix A:")
print(tf.transpose(A))

# Transpose of Matrix B
print("\nTranspose of Matrix B:")
print(tf.transpose(B))
```

```python
# Create a matrix
A = tf.constant([[1, 2, 3],
                 [4, 5, 6]])

print("Matrix A:")
print(A)

# Sum of all elements
print("\nSum of all elements:")
print(tf.reduce_sum(A))

# Sum of each column
print("\nColumn-wise Sum:")
print(tf.reduce_sum(A, axis=0))

# Sum of each row
print("\nRow-wise Sum:")
print(tf.reduce_sum(A, axis=1))

# Mean of all elements
print("\nMean of all elements:")
print(tf.reduce_mean(tf.cast(A, tf.float32)))

# Mean of each column
print("\nColumn-wise Mean:")
print(tf.reduce_mean(tf.cast(A, tf.float32), axis=0))

# Mean of each row
print("\nRow-wise Mean:")
print(tf.reduce_mean(tf.cast(A, tf.float32), axis=1))
```

```python
# Create a 2-D tensor
tensor = tf.constant([[10, 20, 30],
                      [40, 50, 60],
                      [70, 80, 90]])

print("Original Tensor:")
print(tensor)

# Indexing
print("\nFirst Row:")
print(tensor[0])

print("\nSecond Row:")
print(tensor[1])

print("\nElement at Row 2, Column 3:")
print(tensor[1, 2])

# Slicing
print("\nFirst Two Rows:")
print(tensor[:2])

print("\nSecond Column:")
print(tensor[:, 1])

print("\nLast Two Columns:")
print(tensor[:, 1:])
```

```python
# Create a TensorFlow Variable
x = tf.Variable([10, 20, 30])

print("Original Variable:")
print(x)

# Update the variable
x.assign([100, 200, 300])

print("\nAfter assign():")
print(x)

# Add a value to each element
x.assign_add([10, 10, 10])

print("\nAfter assign_add():")
print(x)

# Subtract a value from each element
x.assign_sub([5, 5, 5])

print("\nAfter assign_sub():")
print(x)
```
